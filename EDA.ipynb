```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

# Load the cleaned dataset
df = pd.read_csv("CleanData/himalayan_climbers_clean.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
#Basic information  about the dataset
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nData types:")
print("df.dtypes")
print("\nMissing values:")
print(df.isnull().sum())
```

```python
#Check for duplicates rows
print("Number of duplicates rows:",df.duplicated().sum())
```

```python
#Summary Statisrics for numerical columns
df.describe()
```

```python
#Number of unique value in each column
print("Unique values in each column:")
print(df.nunique().sort_values())
```

```python
#Check value of categorical columns
categoricals_cols = df.select_dtypes(include="object").columns
for col in categoricals_cols:
    print(f"\n---{col}---")
print(df[col].value_counts().head(10))
```

```python
#Check the distribution of target variable
print("Distribution of died:")
print(df["died"].value_counts(dropna=False))

print("\nPercentage distribution:")
print(
    (df["died"].value_counts(normalize=True, dropna=False) * 100)
    .round(2)
)
```

```python

# Visualize target distribution
sns.countplot(data=df, x="died")
plt.title("Distribution of Deaths")
plt.xlabel("Died")
plt.ylabel("Number of records")
plt.show()
```

```python
#Summary statistics of numerical columns
print("Numerical feature summary:")
display(df.describe())
    
```

```python
#Distribution of important numerical features
numerical_cols = df.select_dtypes(include="number").columns.tolist()
df[numerical_cols].hist(
    figsize=(12,8),
    bins=30,
    edgecolor="black"
)
plt.suptitle("Distribution of numerical features")
plt.tight_layout()
plt.show()
```

```python
#Analyze categorical features
categoricals_cols= df.select_dtypes(include=["object","category"]).columns
for col in categoricals_cols:
    print(f"\n----{col}---")
print(df[col].value_counts().head(10))    
```
